```r
### Show how often each tournament seed occurs.
ggplot(tidy_march_madness, aes(x = seed)) +
    geom_histogram(binwidth = 1, center = 0) +
    scale_x_continuous(breaks = 0:16) +
    labs(
        title = "Distribution of Tournament Seeds",
        x = "Tournament seed",
        y = "Number of tournament appearances",
        caption = "Seed 0 marks teams in the 1983 opening round."
    )
```

```r
# Count each tournament finish and calculate its proportion.
finish_proportions <- tidy_march_madness |>
  count(tourney_finish) |>
  mutate(proportion = n / sum(n))

# Plot the proportions as percentages.
ggplot(finish_proportions, aes(x = proportion, y = tourney_finish)) +
  geom_col() +
  scale_x_continuous(labels = scales::label_percent()) +
  labs(
    title = "Distribution of Tournament Finishes",
    x = "Percentage of tournament appearances",
    y = "Tournament finish"
  )
```

This boxplot compares tournament seeds across finish categories. It shows whether teams that advance further tend to have lower seed numbers and how much the seed distributions overlap across different finishes.

```r
# Compare seed distributions across tournament finishes.
ggplot(tidy_march_madness, aes(x = seed, y = tourney_finish)) +
  geom_boxplot() +
  labs(
    title = "Tournament Seeds by Finish",
    x = "Tournament seed",
    y = "Tournament finish",
    caption = "Seed 0 marks teams in the 1983 opening round."
  )
```

This stacked bar chart shows the proportion of each tournament finish within each seed. Comparing the bars shows whether teams with different seeds tend to have different outcomes, which helps assess whether seed could be useful for predicting tournament finish.

```r
# Show the proportions of tournament finishes within each seed.
ggplot(tidy_march_madness,
       aes(x = factor(seed), fill = tourney_finish)) +
  geom_bar(position = "fill") +
  labs(
    title = "Tournament Finish Proportions by Seed",
    x = "Tournament seed",
    y = "Proportion within each seed",
    fill = "Tournament finish",
    caption = "Seed 0 marks teams in the 1983 opening round."
  )
```

Before splitting the data, I’ll remove records with seed 0 because it was a special label for the 1983 opening round.
